# Dictionary

Case Study– Departmental Data Tracker
Instructions:

Create a dictionary department with keys: "dept_name", "faculty", "courses", "budget".

Add and modify items using dictionary methods (update, pop, get, etc.).

Create two sets: online_courses and offline_courses.

Perform union, intersection, difference, and symmetric_difference.
Create a frozenset named permanent_faculty.

Display all results without loops.

```python
# from types import MethodDescriptorType

# Step 1: Department Dictionary

Dictionary={
      "D001": {"dept_name": "CSE", "faculty": "Vinay", "course": "Java", "budget":25000},
      "D002": {"dept_name": "IT", "faculty": "Babu", "course": "Cloud Computing", "budget":30000},
      "D003": {"dept_name": "ECE", "faculty": "Arjun", "course": "Digital Electronics", "budget":20000}
           }

print(f'Dictionary Creation:',Dictionary)


# Step 2: Modify values
 # get method

print(f'Get Dictionary D001:',Dictionary.get("D001"))


print(f'Get Dictionary D002 with their course:',Dictionary["D002"].get("course"))


 # upadate Method
Dictionary["D001"].update({"course": "Python", "budget": 30000})
print(f'Update Dictionary D001 with their course and budget:',Dictionary["D001"])

Dictionary.update({
    "D004": {
        "dept_name": "Mechanical",
        "faculty": "Kumar",
        "course": "Thermodynamics",
        "budget": 28000
},
    "D005": {
        "dept_name": "Civil",
        "faculty": "Priya",
        "course": "Structural Engineering",
        "budget": 32000
    }
    })
print(f'Update Dictionary D004 and D005 details:',Dictionary)

# pop method

print(f'Pop Dictionary D003:',Dictionary.pop("D003"))

print(f'PopItem Dictionary:',Dictionary.popitem())



# Step 3: Set Operations

online_courses = {23,89,45,78,145}
offline_courses ={104,125,145,168,89}

print(f'Union Values:',online_courses.union(offline_courses))
print(f'Intersection Values:',online_courses.intersection(offline_courses))
print(f'Difference Values:',online_courses.difference(offline_courses))
permanent_counts = online_courses.symmetric_difference(offline_courses)
print(f'symmetric_difference Values:',permanent_counts)

# Step 4: Frozenset Example

Frozen_Set_Values = frozenset(permanent_counts)
print(Frozen_Set_Values)
```
