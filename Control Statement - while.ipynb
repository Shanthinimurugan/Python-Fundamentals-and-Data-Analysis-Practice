# Control Statement - while

Write a program to search for a number in a list using while and break. search for a number in the list using while loop when number is found break the loop

```python
#i,len,lis2,seach

list2 = [9,4,8,6,7]
i=0
search = int(input("Enter the number to search:"))
list_length = len(list2)
while i<list_length:
  if list2[i]==search:
    print("Found the element:", search)
    break
  print("This is not an searched element:", list2[i])
  i+=1

if i == list_length:
 print("Number not found")
```
