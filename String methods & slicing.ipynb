# String methods and slicing 

Actitivity 1:  

Use string methods and slicing to:  
Remove unwanted spaces.  
Convert:  
Name → John Doe  
Email → john.doe@company.com  
Extract the Employee ID and print only the numeric part: 0087  

```python
# Step 1: Clean the string

record = " EmployeeName :  john doe "
record = record.strip()
print(record)

Email = "JOHN.DOE@company.COM"
print(Email)

ID =  " emp_0087 "
ID = ID.strip()
print(ID)
```

```python
# Step 2: Extract name

name = record.split(":")
print(name)

name = name[1]
print(name)

name = name.lstrip()
print(name)
```

```python
# Step 3: Extract email

Email = Email.replace("JOHN.DOE@company.COM","john.doe@company.com")
print(Email)
```

```python
# Step 4: Extract employee ID and get numeric part

ID = ID[-4:]
print(ID)
```

```python
# Step 5: Final Output

print("Name=",name,"\nEmail=",Email,"\nID=",ID)
```
