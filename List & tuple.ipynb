# Retail Inventory Manager

```python
# Tuple, Immutable vs Mutable, List Creation
# 1.Create a tuple with fixed product categories like Electronics, Stationery, and Clothing

P_Category = "Electronics", "Stationery", "Clothing"
print(P_Category)
print(type(P_Category))
```

```python
# 2.Create a list of products that can be updated dynamically

Products = ["Laptop", "Smartphone", "Pen", "Pencil", "Shoes", "Watch"]
print(Products)
print(type(Products))
```

```python
# 3.Ask the user to enter a new product and add it to the list
new_product=input("Enter a New Product Name:")
Products.append(new_product)
Products
```

```python
# 4.Ask the user to input the index of a product to remove it from the list
index = int(input("Enter the index of the product to remove:"))
Products.pop(index)
print("Products list after removing an item:")
Products
```

```python
# 5.Display the final list of products and the product categories
print("Product Names:", Products)
print("Products Categories:", P_Category)
```

# Contact Directory

```python
# List Methods and Slicing
# 1.Create a 2D list where each row holds a contact’s name, phone number, and email address
Directory = [["Kala",9865563245,"kala@gmail.com"],
             ["Mala",9798145623,"mala@gmail.com"],
             ["Raj",9001289755,"raj@gmail.com"]]
Directory
```

```python
# 2.Allow the user to add a new contact
name, mobile, email = input("Enter a new conatct details:").split(",")
new_contact = [name, int(mobile), email]
Directory.append(new_contact)
Directory
```

```python
# 3.Remove the last item
Directory.pop()


```

```python
# 4.Get new items from the users using input() and add those items to the list
name = input("Enter your name=")
phone_number = int(input("Enter your Phone Number="))
email = input("Enter your Email=")
Directory.append([name,phone_number,email])
print(Directory)    
```

```python

Directory.insert(2, ["Zohan", 5689451278, "zohan@gmai.com"])
Directory
```

```python

#  Sort the list in ascending order based on any values
Directory.sort()
Directory
```
