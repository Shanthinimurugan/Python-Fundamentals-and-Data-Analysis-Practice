Student Activity 1 :
* Create two variables: height and grade.
* Store these values in two variables, height and grade, which are float and integer datatypes, respectively.
* Next, convert the height and grade variables to a string datatype.

```python
# Create two variables: height and grade.
# Store these values in two variables, height and grade, which are float and integer datatypes, respectively.


height = float(input("Enter the height: "))
grade = int(input("Enter the grade: "))
print("Height:", height, "\n", "Grade:", grade)
print("Height datatype=", type(height), "\n", "Grade datatype=", type(grade))


```

```python
# convert the height and grade variables to a another datatype

_Height = int(height)
print("Float type changed from", height, "to", _Height, type(_Height))

_Grade = str(grade)
print("Int type changed from", grade, "to", _Grade, type(_Grade))
```
