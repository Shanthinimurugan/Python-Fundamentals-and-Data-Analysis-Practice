# User Defined Function

Activity 1:  

Create a simple calculator using user-defined functions.


Create separate functions for the following five mathematical operations:
- Addition
- Subtraction
- Multiplication
- Power (Exponentiation)
- Division

```python
def addition(a, b):
  c = a+b
  return c

def subtraction(a, b):
  d = a-b
  return abs(d)

def multiplication(a,b):
  e = a*b
  return e

def power(a,b):
  f = a**b
  return f


def division(a,b):
  if b == 0:
    return "Cannot divide by zero!"
  g = a/b
  return round(g, 2)

number1 = float(input("Enter the first number: "))
number2 = float(input("Enter the second number: "))
choice = input("Enter your choice (1 to 5): ")

if choice == "1":
  print("Addition Result is", addition(number1, number2))
elif choice == "2":
 print("Subtraction Result is", subtraction(number1, number2))
elif choice == "3":
  print("Multiplication Result is", multiplication(number1, number2))
elif choice == "4":
  print("Power Result is", power(number1, number2))
elif choice == "5":
  print("Division Result is", division(number1, number2))
else:
  print("Invalid Choice")

c = addition(number1, number2)
d = subtraction(number1, number2)
e = multiplication(number1, number2)
f = power(number1, number2)
g = division(number1, number2)

print("\nAll operations Result")
print("Addition:", c)
print("Subtraction:", d)
print("Multiplication:", e)
print("Power:", f)
print("Division:", g)
```
