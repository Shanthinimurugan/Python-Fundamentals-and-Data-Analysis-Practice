# Tuple Creation

🧩 Scenario:  
A school wants to maintain student data and marks for analysis.  

Each student’s personal details are fixed (Name, Roll Number, Course) → stored in a Tuple.  
Each student’s subject marks are modifiable → stored in a List.  
A collection of all students’ marks is stored as a 2D List.  
For advanced organization, a 3D List is used to represent marks for multiple batches.  

```python
# Step 1: Create Tuple for each student (immutable personal info)

Stud1=("Kala",111,"Biology")
Stud2=("Mala",222,"Commerce")
Stud3=("Ajay",333,"ComputerScience")
Stud4=("Dev",444,"Pure Science")
Stud5=("Sindhu",555,"History")
Stud6=("Bala",666,"Economics")
print(Stud1,"\n", Stud2,"\n", Stud3, "\n", Stud4,"\n", Stud5,"\n", Stud6)
```

```python
# Step 2: Create 1D List for marks for 3 students
marks1 = [89,90,85]
marks2 = [56,75,92]
marks3 = [45,65,87]

print(f"{marks1}\n{marks2}\n{marks3}")
```

```python
# Step 3: Combine all marks into a 2D List
Batch1 =[marks1,marks2,marks3]
Batch1
```

```python
# Step 4: Combine all batches (3D List - two different batches)
marks4 = [82, 75, 89]
marks5 = [91, 87, 84]
marks6 = [78, 90, 88]
Batch2 = [marks4, marks5, marks6]

# create another 2d called batch2 and store 3 more students marks
Batches = [Batch1, Batch2]
Batches
```

```python
# Step 5: Display formatted data

print("Students Personal Details:")

print(Stud1)
print(Stud2)
print(Stud3)
print(Stud4)
print(Stud5)
print(Stud6)

print("\nAll 6 students Marks:")
print(f"{marks1}\n{marks2}\n{marks3}\n{marks4}\n{marks5}\n{marks6}")

print("\nBatch 1 Marks:")

print(Batch1)

print("\nBatch 2 Marks:")

print(Batch2)

print("\nAll Batches (3D List):")

print(Batches)
```
